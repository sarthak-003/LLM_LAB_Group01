# Task 2: Linear Regression Using a Multi-Layer Neural Network From Scratch


## 1. Import Libraries


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing

np.random.seed(42)

print("Libraries imported successfully.")
```

## 2. Load the California Housing Dataset


```python
housing = fetch_california_housing(as_frame=True)

df = housing.frame.copy()

print("Dataset shape:", df.shape)
display(df.head())

print("\nAvailable columns:")
print(df.columns.tolist())
```

```python
feature_names = ["MedInc", "HouseAge"]
target_name = "MedHouseVal"

X = df[feature_names].to_numpy(dtype=np.float64)
y = df[target_name].to_numpy(dtype=np.float64).reshape(-1, 1)

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeature statistics:")
display(pd.DataFrame(X, columns=feature_names).describe())

print("\nTarget statistics:")
display(pd.DataFrame(y, columns=[target_name]).describe())
```

## 3. Train-Test Split and Feature Normalization


```python
rng = np.random.default_rng(42)

n_samples = X.shape[0]
indices = rng.permutation(n_samples)

split_index = int(0.80 * n_samples)

train_idx = indices[:split_index]
test_idx = indices[split_index:]

X_train = X[train_idx]
X_test = X[test_idx]
y_train = y[train_idx]
y_test = y[test_idx]

X_mean = X_train.mean(axis=0)
X_std = X_train.std(axis=0)

X_std[X_std == 0] = 1.0

X_train_norm = (X_train - X_mean) / X_std
X_test_norm = (X_test - X_mean) / X_std

print("Training samples:", len(X_train_norm))
print("Testing samples :", len(X_test_norm))

print("\nTraining feature means after normalization:")
print(X_train_norm.mean(axis=0))

print("\nTraining feature standard deviations after normalization:")
print(X_train_norm.std(axis=0))
```

## 4. Define the Neural Network From Scratch


```python
class NeuralNetwork:
    def __init__(self, layer_sizes, seed=42):
        self.layer_sizes = layer_sizes
        self.num_layers = len(layer_sizes) - 1

        rng = np.random.default_rng(seed)

        self.W = []
        self.b = []

        for i in range(self.num_layers):
            fan_in = layer_sizes[i]
            fan_out = layer_sizes[i + 1]

            W = rng.standard_normal((fan_in, fan_out)) * np.sqrt(2.0 / fan_in)
            b = np.zeros((1, fan_out))

            self.W.append(W)
            self.b.append(b)

    @staticmethod
    def relu(Z):

        return np.maximum(0, Z)

    @staticmethod
    def relu_derivative(Z):

        return (Z > 0).astype(np.float64)

    def forward(self, X):
        A = X

        activations = [X]
        pre_activations = []

        for layer in range(self.num_layers):
            Z = A @ self.W[layer] + self.b[layer]
            pre_activations.append(Z)

            if layer < self.num_layers - 1:
                A = self.relu(Z)
            else:
                A = Z

            activations.append(A)

        return A, activations, pre_activations

    def predict(self, X):

        predictions, _, _ = self.forward(X)
        return predictions
```

## 5. Loss Function


```python
def mse_loss(y_true, y_pred):

    return np.mean((y_true - y_pred) ** 2)

def total_loss(y_true, y_pred, weights, l2_lambda=0.0):

    data_loss = mse_loss(y_true, y_pred)

    if l2_lambda > 0:
        reg_loss = sum(np.sum(W ** 2) for W in weights)
        reg_loss = (l2_lambda / (2 * len(y_true))) * reg_loss
    else:
        reg_loss = 0.0

    return data_loss + reg_loss
```

## 6. Manual Backpropagation


```python
def backward(model, X, y, activations, pre_activations, l2_lambda=0.0):
    n = X.shape[0]

    dW = [None] * model.num_layers
    db = [None] * model.num_layers

    y_pred = activations[-1]

    dA = (2.0 / n) * (y_pred - y)

    for layer in reversed(range(model.num_layers)):
        Z = pre_activations[layer]

        if layer == model.num_layers - 1:
            dZ = dA
        else:

            dZ = dA * model.relu_derivative(Z)

        A_prev = activations[layer]

        dW[layer] = A_prev.T @ dZ
        db[layer] = np.sum(dZ, axis=0, keepdims=True)

        if l2_lambda > 0:
            dW[layer] += (l2_lambda / n) * model.W[layer]

        dA = dZ @ model.W[layer].T

    return dW, db
```

## 7. Optimizers


```python
class Optimizer:

    def __init__(self, model, lr=0.001):
        self.lr = lr

    def update(self, model, dW, db):
        raise NotImplementedError

class GradientDescent(Optimizer):

    def update(self, model, dW, db):
        for i in range(model.num_layers):
            model.W[i] -= self.lr * dW[i]
            model.b[i] -= self.lr * db[i]

class Momentum(Optimizer):

    def __init__(self, model, lr=0.001, beta=0.9):
        super().__init__(model, lr)
        self.beta = beta

        self.vW = [np.zeros_like(W) for W in model.W]
        self.vb = [np.zeros_like(b) for b in model.b]

    def update(self, model, dW, db):
        for i in range(model.num_layers):
            self.vW[i] = self.beta * self.vW[i] + dW[i]
            self.vb[i] = self.beta * self.vb[i] + db[i]

            model.W[i] -= self.lr * self.vW[i]
            model.b[i] -= self.lr * self.vb[i]

class Adam(Optimizer):

    def __init__(
        self,
        model,
        lr=0.001,
        beta1=0.9,
        beta2=0.999,
        epsilon=1e-8
    ):
        super().__init__(model, lr)

        self.beta1 = beta1
        self.beta2 = beta2
        self.epsilon = epsilon

        self.t = 0

        self.mW = [np.zeros_like(W) for W in model.W]
        self.vW = [np.zeros_like(W) for W in model.W]

        self.mb = [np.zeros_like(b) for b in model.b]
        self.vb = [np.zeros_like(b) for b in model.b]

    def update(self, model, dW, db):
        self.t += 1

        for i in range(model.num_layers):

            self.mW[i] = self.beta1 * self.mW[i] + (1 - self.beta1) * dW[i]
            self.mb[i] = self.beta1 * self.mb[i] + (1 - self.beta1) * db[i]

            self.vW[i] = self.beta2 * self.vW[i] + (1 - self.beta2) * (dW[i] ** 2)
            self.vb[i] = self.beta2 * self.vb[i] + (1 - self.beta2) * (db[i] ** 2)

            mW_hat = self.mW[i] / (1 - self.beta1 ** self.t)
            mb_hat = self.mb[i] / (1 - self.beta1 ** self.t)

            vW_hat = self.vW[i] / (1 - self.beta2 ** self.t)
            vb_hat = self.vb[i] / (1 - self.beta2 ** self.t)

            model.W[i] -= self.lr * mW_hat / (np.sqrt(vW_hat) + self.epsilon)
            model.b[i] -= self.lr * mb_hat / (np.sqrt(vb_hat) + self.epsilon)
```

## 8. Training Function


```python
def train_model(
    model,
    optimizer,
    X_train,
    y_train,
    X_test,
    y_test,
    epochs=1000,
    l2_lambda=0.0,
    print_every=100
):

    train_losses = []
    test_losses = []

    for epoch in range(1, epochs + 1):

        y_pred, activations, pre_activations = model.forward(X_train)

        train_loss = total_loss(
            y_train,
            y_pred,
            model.W,
            l2_lambda=l2_lambda
        )

        dW, db = backward(
            model,
            X_train,
            y_train,
            activations,
            pre_activations,
            l2_lambda=l2_lambda
        )

        optimizer.update(model, dW, db)

        train_losses.append(train_loss)

        test_pred = model.predict(X_test)
        test_loss = mse_loss(y_test, test_pred)
        test_losses.append(test_loss)

        if print_every and (epoch == 1 or epoch % print_every == 0):
            print(
                f"Epoch {epoch:4d}/{epochs} | "
                f"Train Loss: {train_loss:.6f} | "
                f"Test MSE: {test_loss:.6f}"
            )

    return {
        "train_loss": np.array(train_losses),
        "test_loss": np.array(test_losses)
    }
```

## 9. Helper Function for Running an Experiment


```python
def create_optimizer(name, model, lr):

    name = name.lower()

    if name == "gd":
        return GradientDescent(model, lr=lr)

    if name == "momentum":
        return Momentum(model, lr=lr, beta=0.9)

    if name == "adam":
        return Adam(model, lr=lr)

    raise ValueError(f"Unknown optimizer: {name}")

def run_experiment(
    optimizer_name,
    learning_rate,
    epochs=1000,
    architecture=[2, 5, 3, 1],
    l2_lambda=0.0,
    seed=42,
    verbose=False
):

    model = NeuralNetwork(architecture, seed=seed)

    optimizer = create_optimizer(
        optimizer_name,
        model,
        learning_rate
    )

    history = train_model(
        model,
        optimizer,
        X_train_norm,
        y_train,
        X_test_norm,
        y_test,
        epochs=epochs,
        l2_lambda=l2_lambda,
        print_every=100 if verbose else 0
    )

    predictions = model.predict(X_test_norm)
    test_mse = mse_loss(y_test, predictions)

    return model, history, predictions, test_mse
```

# 10. Experiment 1 — Basic Gradient Descent


```python
gd_results = {}

for lr in [0.01, 0.001]:
    print(f"\nTraining GD with learning rate = {lr}")

    model, history, predictions, test_mse = run_experiment(
        optimizer_name="gd",
        learning_rate=lr,
        epochs=1000,
        architecture=[2, 5, 3, 1],
        seed=42,
        verbose=True
    )

    gd_results[lr] = {
        "model": model,
        "history": history,
        "predictions": predictions,
        "test_mse": test_mse
    }

    print(f"Final Test MSE: {test_mse:.6f}")
```

```python
plt.figure(figsize=(10, 6))

for lr, result in gd_results.items():
    plt.plot(
        result["history"]["train_loss"],
        label=f"GD, lr={lr}"
    )

plt.xlabel("Epoch")
plt.ylabel("Training MSE")
plt.title("Gradient Descent: Learning Rate Comparison")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Interpretation


# 11. Experiment 2 — Compare GD, Momentum and Adam


```python
optimizer_results = {}

for optimizer_name in ["gd", "momentum", "adam"]:
    print(f"\nTraining {optimizer_name.upper()} with lr=0.001")

    model, history, predictions, test_mse = run_experiment(
        optimizer_name=optimizer_name,
        learning_rate=0.001,
        epochs=1000,
        architecture=[2, 5, 3, 1],
        seed=42,
        verbose=True
    )

    optimizer_results[optimizer_name] = {
        "model": model,
        "history": history,
        "predictions": predictions,
        "test_mse": test_mse
    }

    print(f"Final Test MSE: {test_mse:.6f}")
```

```python
plt.figure(figsize=(10, 6))

for optimizer_name, result in optimizer_results.items():
    plt.plot(
        result["history"]["train_loss"],
        label=optimizer_name.upper()
    )

plt.xlabel("Epoch")
plt.ylabel("Training MSE")
plt.title("Optimizer Convergence Comparison")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
optimizer_table = pd.DataFrame([
    {
        "Optimizer": name.upper(),
        "Learning Rate": 0.001,
        "Final Train MSE": result["history"]["train_loss"][-1],
        "Test MSE": result["test_mse"]
    }
    for name, result in optimizer_results.items()
])

optimizer_table = optimizer_table.sort_values("Test MSE").reset_index(drop=True)

display(optimizer_table)
```

## 12. Learning Rate + Optimizer Summary


```python
all_results = {}

for optimizer_name in ["gd", "momentum", "adam"]:
    for lr in [0.01, 0.001]:

        key = f"{optimizer_name.upper()}_lr_{lr}"

        print(f"Running {key} ...")

        model, history, predictions, test_mse = run_experiment(
            optimizer_name=optimizer_name,
            learning_rate=lr,
            epochs=1000,
            architecture=[2, 5, 3, 1],
            seed=42,
            verbose=False
        )

        all_results[key] = {
            "optimizer": optimizer_name.upper(),
            "lr": lr,
            "model": model,
            "history": history,
            "predictions": predictions,
            "test_mse": test_mse
        }

print("All experiments completed.")
```

```python
summary_table = pd.DataFrame([
    {
        "Experiment": key,
        "Optimizer": result["optimizer"],
        "Learning Rate": result["lr"],
        "Final Train MSE": result["history"]["train_loss"][-1],
        "Test MSE": result["test_mse"]
    }
    for key, result in all_results.items()
])

summary_table = summary_table.sort_values("Test MSE").reset_index(drop=True)

display(summary_table)
```

```python
plt.figure(figsize=(12, 7))

for key, result in all_results.items():
    plt.plot(
        result["history"]["train_loss"],
        label=key
    )

plt.xlabel("Epoch")
plt.ylabel("Training MSE")
plt.title("Learning Rate and Optimizer Comparison")
plt.legend(fontsize=8)
plt.grid(alpha=0.3)
plt.show()
```

# 13. Evaluation — Predicted vs Actual Values


```python
best_key = min(all_results, key=lambda k: all_results[k]["test_mse"])
best_result = all_results[best_key]

best_predictions = best_result["predictions"]

print("Best experiment:", best_key)
print("Best test MSE:", best_result["test_mse"])
```

```python
plt.figure(figsize=(8, 8))

plt.scatter(
    y_test.ravel(),
    best_predictions.ravel(),
    alpha=0.25,
    s=12
)

minimum = min(y_test.min(), best_predictions.min())
maximum = max(y_test.max(), best_predictions.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--",
    linewidth=2
)

plt.xlabel("Actual MedHouseVal")
plt.ylabel("Predicted MedHouseVal")
plt.title(f"Predicted vs Actual — {best_key}")
plt.grid(alpha=0.3)
plt.show()
```

## 14. Final Test MSE


```python
final_mse = mse_loss(y_test, best_predictions)

print(f"Final Test MSE = {final_mse:.6f}")

rmse = np.sqrt(final_mse)
print(f"Test RMSE      = {rmse:.6f}")
```

```python
no_reg_model, no_reg_history, no_reg_predictions, no_reg_mse = run_experiment(
    optimizer_name="adam",
    learning_rate=0.001,
    epochs=1000,
    architecture=[2, 5, 3, 1],
    l2_lambda=0.0,
    seed=42,
    verbose=False
)

l2_model, l2_history, l2_predictions, l2_mse = run_experiment(
    optimizer_name="adam",
    learning_rate=0.001,
    epochs=1000,
    architecture=[2, 5, 3, 1],
    l2_lambda=0.001,
    seed=42,
    verbose=False
)

regularization_table = pd.DataFrame({
    "Model": [
        "Without L2",
        "With L2 (lambda=0.001)"
    ],
    "Final Train Loss": [
        no_reg_history["train_loss"][-1],
        l2_history["train_loss"][-1]
    ],
    "Test MSE": [
        no_reg_mse,
        l2_mse
    ]
})

display(regularization_table)
```
